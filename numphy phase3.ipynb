Aggregate function

```python
import numpy as np
arr=np.array([10,20,30,40,50])
print("sum:",np.sum(arr))
print("max:",np.max(arr))
print("min:",np.min(arr))
print("count:",arr.size)
print("cumulative sum:",np.cumsum(arr))
print("cumulative product:",np.cumprod(arr))
```

Statistical function

```python
import numpy as np
import statistics as stats
sales=np.array([200,180,150,250,270,300,250,230])
print("mean:",np.mean(sales))
print("median:",np.median(sales))
print("mode:",stats.mode(sales))
print("standard deviation:",np.std(sales))
```

```python
arr=np.array([19,23,17,22,18])
print(np.std(arr))
arr1=np.array([15,20,30,10,50])
print(np.std(arr1))
```
