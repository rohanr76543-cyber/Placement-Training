GroupBy,MergeOrJoin

```python
import pandas as pd
```

```python
exp = pd.read_excel("subdata_pandas/expense.xlsx")
exp
```

```python
exp.groupby("Category")['Amount'].sum()
```

```python
exp.groupby('Payment Mode')["Amount"].agg(['sum'])
```

```python
exp.groupby("Category")['Amount'].agg(['sum'])
```

```python
exp.groupby('Payment Mode')["Amount"].agg(['sum','count'])
```

```python
exp.groupby("Category")['Amount'].agg(['sum','mean','max','min'])
```

```python
exp.groupby(['Payment Mode','Category'])['Amount'].agg(['sum','count'])
exp.groupby(['Category','Payment Mode'])['Amount'].agg(['sum','count'])
```

```python
exp.groupby("Category")['Amount'].sum().sort_values(ascending=False)
```

##MergeOrJoin

```python
customer=pd.read_csv("subdata_pandas/customer1.csv")
customer
payment=pd.read_csv("subdata_pandas/payment.csv")
payment
```

#Inner join

```python
pd.merge(left=customer,right=payment,on='customerid',how='inner')
```
