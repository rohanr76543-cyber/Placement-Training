##introduction to Matplotlin

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
cources=['Python','Java','C++','C#','JavaScript']
enrolled=[200,150,300,250,400]
plt.bar(cources,enrolled,color='green',width=0.4)
plt.show()
```

```python
cource_enroll={
    "cources":['java','python','c','c++'],
    "enrolled":[25,15,18,10]
}
cource_df=pd.DataFrame(cource_enroll)
plt.figure(figsize=(6,5))
bars=plt.bar(cource_df['cources'],cource_df['enrolled'],color=['#C736BD','#0BDECC','#0BDE5C','#D7DE0B'],edgecolor='black',linewidth=0.4)
#creating title and labels
plt.title('Course Enrolled Student Overview',fontsize=15,fontweight='bold',pad=5)
plt.xlabel('Courses',fontsize=12)
plt.ylabel('Number of Students',fontsize=12)
plt.bar_label(bars,padding=2,fontweight='bold',fontsize=10)

#Improve
plt.ylim(0,30)
plt.grid(axis='y',linestyle='--',alpha=0.7)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```

## line plot

```python
weekly_sales={
    "days":['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'],
    'first_week':[35000,33000,38000,32000,39000,33000,45000],
    'second_week':[32000,35000,42000,37000,34000,43000,48000]
}
sales=pd.DataFrame(weekly_sales)
plt.figure(figsize=(8,5))
#------------first week sales-------------
plt.plot(sales['days'],sales['first_week'],color='red',ls="--",marker="o",label='First Week Sales')

#------------second week sales-------------
plt.plot(sales['days'],sales['second_week'],color='green',marker="o",label='Second Week Sales')

#create title and labels
plt.title('Weekly Sales Overview',fontsize=15,fontweight='bold',pad=5)
plt.xlabel('Days',fontsize=12)
plt.ylabel('Sales amount per day',fontsize=12)
plt.legend()
plt.tight_layout()
plt.show()
```

## Scatter plot

```python
student_age= {
    "age":[20,22,21,20,21,22,23,22,20,24,22,21,21,21,20,23,23,24,25,25,32],
    "marks":[490,480,470,460,450,440,430,490,480,470,460,450,440,430,490,480,470,460,450,440,430]
}
std_df=pd.DataFrame(student_age)
plt.figure(figsize=(8,5))
plt.scatter(std_df['age'],std_df['marks'],color='red',marker='o',s=100,edgecolor='yellow')
#create title and labels
plt.title('Student Ages and Marks',fontweight='bold',fontsize=15)
plt.xlabel('Age',fontsize=12)
plt.ylabel('Marks',fontsize=12)
plt.legend(['Student Data'],loc='upper right')
plt.show()  
```

```python
weekly_sales={
    'Brand':['Oneplus','Apple','Oppo','Vivo','Samsung'],
    'Sales':[23,10,19,32,25]
}
sales_df=pd.DataFrame(weekly_sales)
plt.pie(
sales_df['Sales'],labels=sales_df['Brand'],autopct='%1.1f%%',shadow=True,explode=[0,0,0.1,0,0],startangle=30,pctdistance=0.8
)
plt.title("Weekly Sales Overview",fontsize=15,fontweight='bold')
plt.legend(loc='upper right',bbox_to_anchor=(1.25,1))
plt.show()
```

```python
data=pd.read_excel("expense.xlsx")
exp=pd.DataFrame(data)
exp_group=exp.groupby('Category')['Amount'].sum()
plt.pie(exp_group.values,labels=exp_group.index,autopct="%.2f%%",explode=[0,0.1,0,0,0],shadow=True)
plt.show()
```

Histogram

```python
temp_celsius=[28,29,30,31,31,30,29,28,27,26,25,24,24,24]
plt.hist(temp_celsius,bins=5,color='green',alpha=0.7,edgecolor='black')
#creating title and labels
plt.title('Distribution of today hourly temperatures',fontsize=15,fontweight='bold',pad=5)
plt.xlabel('Temperature (°C)',fontsize=12)
plt.ylabel('Frequency',fontsize=12)
plt.grid(axis='y',linestyle='--',alpha=0.7)
plt.show()
```

Stack plot

```python
# Weekly Sales Data (in units)
weekly_sales = {
    "Days": ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"],
    "Week1": [5, 10, 30, 20, 35, 60, 80],
    "Week2": [30, 60, 30, 70, 80, 50, 30],
    "Week3": [10, 40, 50, 23, 69, 82, 43]
}
sales=pd.DataFrame(weekly_sales)
plt.stackplot(sales['Days'],sales['Week1'],sales['Week2'],sales['Week3'],labels=['Week1','Week2','Week3'],colors=["#730CF1","#DE510B",'#0BDE5C'])
plt.legend(loc='upper left')
plt.show()
```
