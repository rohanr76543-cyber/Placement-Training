```python
cdata.duplicated()
cdata[cdata.duplicated()]
cdata['EEID'].duplicated()
cdata[cdata['EEID'].duplicated()]
cdata.drop_duplicates()
cdata['EEID'].drop_duplicates()
```

```python
cdata.isnull()
edata=pd.read_csv("subdata_pandas/employee.csv")
edata.isnull().sum()
cdata.isnull().sum()
```

```python
cdata['salary'].isnull().sum()
```

```python
#Replace NaN with any value
cdata.replace(np.nan,"Insufficient")
#Replace NaN with 30000 in salary
cdata['salary']=cdata['salary'].replace(np.nan, 30000)
#replaces NaN salary with average
mean=np.mean(cdata['salary'])
mean
#M1
#cdata['salary'].replace(np)
#cdata['salary']=cdata['salary'].replace(np.nan,mean)
#cdata
#M2
cdata['salary']=cdata['salary'].fillna(mean)
cdata
#forward fill(prev row)
cdata.ffill()
#backward fill(next row)
cdata.bfill()
```
